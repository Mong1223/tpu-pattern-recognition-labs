```python
import numpy as np
from sklearn import svm
from sklearn import metrics
```

```python
from sklearn.datasets import load_breast_cancer
data = load_breast_cancer()
```

```python
features = data.data
target = data.target
```

```python
features.shape
```

```python
target.shape
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(features, target, test_size=0.3,random_state=123) 
```

```python
for i in range(1, 20):
  clf = svm.SVC(C=i, kernel='linear')
  clf.fit(X_train, y_train)
  print()
  print("Accuracy for SVC when using linear kernal with C = ",i, " ",metrics.accuracy_score(y_test, clf.predict(X_test)))
```

```python
for i in range(1, 20):

  #Create a svm Classifier
  clf = svm.SVC(C=i, kernel='rbf')# rbf Kernel

  #Train the model using the training sets
  clf.fit(X_train, y_train)
  
  print()
  #Testing the model using testing set
  print("Accuracy for SVC when using rbf kernal with C = ",i, " ",metrics.accuracy_score(y_test, clf.predict(X_test)))
```

```python
for i in range(1, 20):

  #Create a svm Classifier
  clf = svm.SVC(C=i, kernel='poly')# poly Kernel

  #Train the model using the training sets
  clf.fit(X_train, y_train)
  
  print()
  #Testing the model using testing set
  print("Accuracy for SVC when using poly kernal with C = ",i, " ",metrics.accuracy_score(y_test, clf.predict(X_test)))
```

```python
from sklearn.model_selection import GridSearchCV

# Set the parameters by cross-validation
tuned_parameters = [{'C': range(1, 10),
                                          'degree': range(3, 5),
                     'coef0': np.arange(0, 1, 0.1),
    'kernel': ['poly' , 'rbf', 'linear']

                     }]

clf = GridSearchCV(
        svm.SVC(), tuned_parameters
    )

clf.fit(X_train, y_train)

print("Best parameters set found on development set:")
print()
print(clf.best_params_)

print("Accuracy:",metrics.accuracy_score(y_test, clf.predict(X_test)))

```
